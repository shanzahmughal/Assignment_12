```python
import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv(r"C:\Users\36118\Downloads\teaching_prices_messy.csv")
# 2. Drop duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

# 3. Clean string columns (trim whitespace)
string_cols = ['item', 'city', 'unit', 'data_label']
for col in string_cols:
    df[col] = df[col].astype(str).str.strip()

# 4. Clean date column
df['date'] = pd.to_datetime(df['date'])

# 5. Clean price column: remove commas, convert to float
df['price'] = df['price'].astype(str).str.replace(',', '', regex=False)
df['price'] = pd.to_numeric(df['price'], errors='coerce')

# 6. Replace sentinel values/outliers (e.g., 9999 or clear typos) with NaN
# Option A: Replace specific sentinel values
df.loc[df['price'] == 9999, 'price'] = np.nan

# Option B (Optional): Impute missing price values by item and city median/mean
df['price'] = df.groupby(['item', 'city'])['price'].transform(lambda x: x.fillna(x.median()))

# 7. Save the cleaned dataset
df.to_csv('teaching_prices_cleaned.csv', index=False)
print("Dataset successfully cleaned and saved to 'teaching_prices_cleaned.csv'!")
```
